```python
import matplotlib.pyplot as plt
import numpy as np

countries = ("United States", "China", "Japan", "Australia", "France (Host)")
medal_counts = {
    'Gold': (40, 40, 20, 18, 16),
    'Silver': (44, 27, 12, 19, 26),
    'Bronze': (42, 24, 13, 16, 22),
}

x = np.arange(len(countries))  # label konumları
width = 0.25  # barların genişliği
multiplier = 0

fig, ax = plt.subplots(figsize=(10, 5), layout='constrained')

# Her bir madalya kategorisi için barları çizdiriyoruz
for attribute, measurement in medal_counts.items():
    offset = width * multiplier
    rects = ax.bar(x + offset, measurement, width, label=attribute)
    ax.bar_label(rects, padding=3)
    multiplier += 1

ax.set_ylabel('Medal Count')
ax.set_title('Olympic Medal Counts by Country')
ax.set_xticks(x + width, countries)
ax.legend(loc='upper right', ncols=3)
ax.set_ylim(0, 52)

plt.show()
```
